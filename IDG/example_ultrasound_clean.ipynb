## Background information
Integrated Decision Gradients (IDG) (Walker et al. 2023) and similar gradient-based attribution methods (e.g., Integrated Gradients (IG) (Sundararajan et al. 2017), 
Layer-wise Integrated Gradients (LIG) (Suzuki et al. 2026), and Generalized Integrated Gradients (GIG) (Merrill et al. 2019)) are explainability techniques used 
to reveal which parts of an input image were significant when a computer vision (CV) model was classifying the image.

Of these techniques, IDG is the method that appears to yield the most relevant results with the least the amount of attribution placed on irrelevant parts of ultrasound images.

## Combined attribution map
Each attribution map relates to a single gesture. A ViT trained on any one attribution map will be best at classifying that gesture. A multi-channel ViT model will be required to classify images that belong to any class, with each channel responsible for predicting its trained gesture.

In order to train a single ViT model to identify any gesture, we can try training the ViT on an attention map built from attribution maps from all gestures.

The pixel values in integrated gradients can be positive or negative, with the magnitude reflecting the contribution of that pixel to the prediction of the gesture, either via an increase (positive) or decrease (negative) in the prediction score matching the test class.

Therefore the absolute value for each pixel in the attribution map reflects the significance of that pixel to the prediction of the gesture class. While a modified version of the integrated gradients algorithm might be used to capture the significance of each pixel to the prediction of _any_ gesture, we should be able to capture equivalent information by taking a sum or an average of pixel values in all (absolute value) attribution maps. The idea here is that a low magnitude value for a pixel on class A's attribution map doesn't diminish that pixel's significance for class B, and so absolute values should be added rather than the raw positive and negative values (which might cancel each other out in a way that doesn't reflect the information we're trying to capture). Further, a negative value for a pixel on class A's attribution map represents a _reduction_ in the prediction for class A, which may be accompanied by an _increase_ in the prediction for other classes, and in this case the values from this pixel from two or more attribution maps would tend to cancel, resulting in a combined attribution map that represents the opposite of what we want.

## Attention bias map with threshold
There are a number of ways in which noise or errors can show up in the ultrasound data:
- movement of the ultrasound probe relative to the axes of the arm
    - could result in translations of pixel values from their "true" position, if we are strictly measuring the flexion and extension of muscles and tendons
    - could result in pixels close to the periphery of the scanning region being counted as significant, if the formation of the gestures by the subject resulted in the scanning region moving laterally in a way that was consistent for any given gesture.
- movement of the ultrasound probe along the skin, or changes in the ultrasound medium over the course of data collection
    - could result in signal values near the surface of the skin being counted as positive indicators
- ultrasound echoes
    - ultrasound sensing is based on an ultrasonic signal reflecting off structures of various densities and the return signal being detected. There is an effect in ultrasound images where part of the initial outgoing signal can reflect off of multiple structures, or the outgoing wave partially reflects off one structure, and the remaining wave reflects off another. This results in "echoes" ﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿﷿ bright patches further from the emitter on the ultrasound image with a similar shape and horizontal position to (brighter) shapes closer to the emitter.

It appears that the attention bias map may exhibit one or more of these effects, as evidenced by the bright region along the sides and top of the scanning region, and the bright horizontal lines toward the bottom of the scanning region.

We may be able to bias the ViT toward focusing on the muscles and tendons themselves, by applying a threshold or a filter to the raw map of sums, such that the regions below a certain brightness (the brightest parts of the map appear to be where we expect the muscles and tendons to be) are de-prioritized.

```python
# coarse_attribution will be a 10x10 or 16x16 array of floats, each representing a function
# of the attribution values in the attribution map for each patch (number of patches matches the ViT).



coarse_attribution_all_sums = [[0.0] * patches_per_dimension for _ in range(patches_per_dimension)]

# get the attribution scores for all pixels in each patch and sum them
for i in range(patches_per_dimension):
    for j in range(patches_per_dimension):
        # coarse_attribution_all.append(np.sum(idg_norm[i*10:(i+1)*10, i*10:(i+1)*10]))
        row_start: int = patch_lines[i]
        row_end: int = patch_lines[i + 1]
        col_start: int = patch_lines[j]
        col_end: int = patch_lines[j + 1]
        # print(f"fetching attribution scores from ({row_start}, {col_start}) to ({row_end}, {col_end})")
        attrib_map_range = normalized_average_attribution_map[row_start:row_end, col_start:col_end]
        coarse_attribution_all_sums[i][j] = np.sum(attrib_map_range)

# print(f"coarse_attribution_all:\n{coarse_attribution_all}")
coarse_attribution_all_sums_array = np.array(coarse_attribution_all_sums)

coarse_attribution_all_sums_max: float = np.max(coarse_attribution_all_sums_array)
coarse_attribution_all_sums_min: float = np.min(coarse_attribution_all_sums_array)
print(f"coarse_attribution_all_sums_max: {coarse_attribution_all_sums_max}")
print(f"coarse_attribution_all_sums_min: {coarse_attribution_all_sums_min}")

# normalize coarse_attribution_all
coarse_attribution_all_sums_normalized = coarse_attribution_all_sums_array / (coarse_attribution_all_sums_array.max() + 1e-12)

coarse_attrib_resized = cv2.resize(coarse_attribution_all_sums_normalized, (224, 224))

plt.imshow(coarse_attribution_all_sums_normalized, cmap='inferno')
plt.colorbar()
plt.show()
# 224 / 16 == 14

fig_attn_bias, ax = plt.subplots(1,1, figsize=(fig_width, fig_height))
im1 = ax.imshow(avg_training_image, cmap="gray")
fig_attn_bias.colorbar(mappable=im1, ax=ax)
im2 = ax.imshow(coarse_attrib_resized, alpha=0.5, cmap='inferno')
fig_attn_bias.colorbar(mappable=im2, ax=ax)
plt.show()

```

```python
# apply a threshold

attrib_file_datetime: str = datetime.now().strftime("%Y%m%d_%H%M%S")

attrib_maps_folder_path: str = "../results/attribution_maps/"
attn_maps_folder_path: str = "../results/attention_maps/"

high_pass_threshold: float = 0.45
coarse_attrib_sums_high_pass = np.where(coarse_attribution_all_sums_normalized > high_pass_threshold, coarse_attribution_all_sums_normalized, 0)

plt.imshow(coarse_attrib_sums_high_pass, cmap='inferno')
plt.title(f"High pass (>{high_pass_threshold}) filtered coarse attribution map")
plt.show()
```

```python
# config
save_combo_attrib_maps: bool = True

# apply a sigmoid function
transition_width: float = 0.15 # w
transition_value: float = 0.65 # x0

transition_width_text: str = f"{str(transition_width).replace('.', 'p')}"
transition_value_text: str = f"{str(transition_value).replace('.', 'p')}"

coarse_attrib_sums_sigmoid = 1 / (1 + np.exp(-(coarse_attribution_all_sums_normalized - transition_value)/transition_width))

coarse_attrib_sums_sigmoid_normalized = coarse_attrib_sums_sigmoid / np.max(coarse_attrib_sums_sigmoid)

fig_sigmoid, axs_sigmoid = plt.subplots(1,2, figsize=(12, 5))
im0 = axs_sigmoid[0].imshow(coarse_attribution_all_sums_normalized, cmap='inferno')
axs_sigmoid[0].set_title(f"Coarse attribution sums map")
divider = make_axes_locatable(axs_sigmoid[0])
cax = divider.append_axes('right', size='5%', pad=0.1)
fig_sigmoid.colorbar(im0, cax=cax, orientation='vertical')

im2 = axs_sigmoid[1].imshow(coarse_attrib_sums_sigmoid_normalized, cmap='inferno')
axs_sigmoid[1].set_title(f"Sigmoid (x0={transition_value}, w={transition_width}) filtered attribution map", wrap=True)
divider = make_axes_locatable(axs_sigmoid[1])
cax = divider.append_axes('right', size='5%', pad=0.1)
fig_sigmoid.colorbar(im2, cax=cax, orientation='vertical')

plt.tight_layout()
plt.show()

attrib_map_all_filename: str = f"attrib_map_combo_norm_{attrib_file_datetime}"
attrib_map_combo_filename: str = os.path.join(attrib_maps_folder_path, attrib_map_all_filename)
if save_combo_attrib_maps:
    np.save(attrib_map_combo_filename, coarse_attribution_all_sums_normalized)

attrib_map_sigmoid_filename: str = f"attrib_map_combo_sigmoid_x0={transition_value_text}_w={transition_width_text}_{attrib_file_datetime}"
attrib_map_sigmoid_filepath: str = os.path.join(attrib_maps_folder_path, attrib_map_sigmoid_filename)
if save_combo_attrib_maps:
    np.save(attrib_map_sigmoid_filepath, coarse_attrib_sums_sigmoid)
```
